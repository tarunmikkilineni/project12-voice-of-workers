## Research questions

**RQ1.** How do digital labour platforms combine reason-giving, human review, response deadlines, and corrective remedies into distinct configurations of documented worker-voice mechanisms?

**RQ2.** How does the breadth of platforms’ documented managerial discretion compare with workers’ corresponding procedural protections across deactivation, payment, ratings, and work-allocation decisions?

**RQ3.** To what extent are platforms’ stated commitments to worker appeals corroborated, qualified, or contradicted by contemporaneous Fairwork evidence, and how do these patterns vary across procedural safeguards?

**RQ4.** What documented procedural barriers constrain access to worker appeals, and how do platforms preserve or restrict access to review after account suspension or deactivation?

**RQ5.** How do platform governance documents distinguish between workers’ rights to challenge individual decisions and their rights to participate through representatives in reviewing or revising decision-making rules?

## Step by step methodology

- **M1 — Fix scope and dates:** preregister the platform–jurisdiction–period unit, freeze date, domain coverage, inclusion/exclusion criteria and temporal matching windows before outcome coding. Start with the 2023 Fairwork cloudwork cohort; expand only with explicitly comparable reports. Do not target a sample size by selecting on outcomes.
- **M2 — Build and freeze the corpus:** register worker-facing terms, deactivation, payment/dispute policies and platform-specific Fairwork passages. Use immutable Git commit URLs for archived policies. Save source dates, valid-from/valid-to dates, jurisdiction, retrieval date, SHA256 and text/page locators. A 2026 policy cannot establish a 2023 policy commitment. Record exclusions, missing documents and coverage limits.
- **M3 — Operationalize constructs:** use the codebook below, piloted on 2–3 cases. Keep present, not stated, not applicable and unresolved distinct. Do not infer human review from generic support or infer contradiction from a Fairwork point not awarded. Use EU Directive Articles 10–11 as a conceptual benchmark, not a retrospective global compliance test; representation also requires its own governance coding.
- **M4 — Independently code and adjudicate:** two people independently code at least 25% of cases, selected by seeded sampling across jurisdictions/categories when available. Retain original codes, evidence and coder IDs; report agreement and Cohen kappa by variable. Resolve disagreements with written rationales in a separate final dataset. Report sparse/degenerate kappa as undefined.
- **M5 — Validate and triangulate:** require unique keys, complete domain grids, evidence locators, reviewed document scope for not-stated codes, explicit inapplicability rationale, valid dates and source types. Fairwork comparisons are manually assessed against the same safeguard and domain. Separate corroboration, qualification, contradiction and no discussion; exclude temporally unmatched pairs in primary RQ3 analyses.
- **M6 — Analyze configurations and asymmetry:** enumerate exact four-safeguard signatures by domain (RQ1); compare four documented powers with four core protections on 0–1 breadth scales using their own applicable denominators (RQ2). Count not stated as zero documented presence, never proof of real-world absence. Keep raw components and denominators visible.
- **M7 — Analyze independent evidence and access:** triangulate only positive policy commitments and display report silence separately (RQ3). Compare exact temporal overlap with 90/365-day windows. Map barriers and post-deactivation off-account routes (RQ4); distinguish individual appeal, representative assistance, consultation and rule influence (RQ5).
- **M8 — Check robustness and limits:** repeat core breadth/configuration summaries on fully applicable domains, report document coverage and unresolved items, and compare temporal windows. Preserve the finite purposive corpus interpretation: no population generalization, causal effects, wellbeing claims or unsupported significance tests. Differences may reflect disclosure, jurisdiction, document availability or period.
- **M9 — Export and audit:** display every figure/table inline; save figures as vector PDFs and preview PNGs, tables as CSVs, supporting plot data as CSVs, captions, coding audit, input copies and environment metadata. Create an output manifest with hashes and a ZIP. Interpret empirical patterns only after reading the generated evidence tables; do not automatically label an expected pattern as a finding.

```python
# 2 Create input schemas and a coding codebook
# Methodology: M2–M4. Research link: All RQs.
SCHEMAS = {
'cases':['platform_id','platform','jurisdiction','work_category','period_start','period_end','included','exclusion_reason'],
'documents':['doc_id','platform_id','source_type','document_type','title','url','local_path','accessed_date','valid_from','valid_to','sha256'],
'evidence':['evidence_id','doc_id','locator','excerpt_or_scope'],
'codes':['platform_id','domain','variable','value','evidence_id','coder_id','rationale'],
'fairwork_matches':['platform_id','domain','safeguard','status','evidence_id','policy_doc_id','fairwork_doc_id','rationale'],
'independent_codes':['platform_id','domain','variable','value','evidence_id','coder_id','rationale']}
for name,cols in SCHEMAS.items(): pd.DataFrame(columns=cols).to_csv(OUT/'templates'/f'{name}.csv',index=False)
definition = {
'notice':'Specific process for communicating an adverse decision.',
'reasons':'Specific reasons for the disputed decision are promised.',
'human_review':'A person can reconsider the decision; generic support is insufficient.',
'reviewer_clarity':'Responsible review role or team is identifiable.',
'deadline':'A bounded platform response or review time is promised; vague promptness is insufficient.',
'access_route':'A specified contact channel or sequence for challenging a decision.',
'representation':'Representative, union or support person can assist an individual appeal.',
'remedy':'Correction, payment, reinstatement or another specified corrective outcome is possible.',
'unilateral_action':'Platform explicitly reserves power to impose the decision without prior worker agreement.',
'broad_grounds':'Open-ended residual grounds such as sole discretion, any reason or other unspecified grounds.',
'withhold_explanation':'Express power to withhold reasons, not mere silence about giving reasons.',
'final_discretion':'Express final or unreviewable platform decision-making authority.',
'login_required':'Appeal route expressly requires an active logged-in account.',
'short_filing_window':'Express appeal filing limit of 14 calendar days or fewer; threshold fixed before coding.',
'fee_required':'Worker must pay a fee to access the review route.',
'language_restriction':'Explicit restriction on languages available for the appeal route.',
'mandatory_external_route':'Internal review bypassed or foreclosed by a mandatory external dispute route.',
'limited_review_scope':'Explicit limits on decisions or grounds that a reviewer may consider.',
'off_account_route':'Appeal channel explicitly usable without an active account.',
'individual_challenge':'Documented entitlement to challenge an individual decision in at least one domain.',
'representative_assistance':'Explicit ability to obtain representative assistance in individual challenges.',
'rule_consultation':'Representatives can be consulted on rules or automated decision systems.',
'rule_influence':'Explicit representative negotiation, consent, co-decision or binding revision rights; consultation alone is insufficient.'}
codebook=pd.DataFrame([{'variable':k,'definition':v,'scope':'platform' if k in GOV else 'each decision domain'} for k,v in definition.items()])
codebook.to_csv(OUT/'templates'/'codebook.csv',index=False)
display(codebook)
print('Values: present = explicit evidence; not_stated = searched corpus silent; not_applicable = reasoned exclusion; unresolved = awaiting adjudication.')
print('documents.source_type: policy, fairwork, or benchmark. Validity intervals must be verified; do not substitute retrieval date.')
print('For a report, valid_from/to delimit its assessment period, not presumed policy validity. Use the best supported period and disclose uncertainty.')
print('fairwork_matches: one row per PRESENT safeguard claim; not_discussed records a scoped search, not contradiction. policy_doc_id must be the claim source.')
print('Optional independent_codes retains two independent coders; codes.csv is adjudicated. Add original disagreements and resolution rationale to rationale.')
shutil.make_archive(str(BASE/'coding_templates'),'zip',OUT/'templates')
display(FileLink(str(BASE/'coding_templates.zip')))
```

### 4 Retrieve frozen sources and surface candidate passages

**Methodology:** M2–M3. **Research link:** All RQs.

Optional retrieval uses the registered exact URLs or supplied local files, records hashes, and extracts PDF/text passages. Candidate snippets are discovery aids only: a person must read the full surrounding policy and enter verified evidence. HTML/PDF extraction does not establish any safeguard.


```python
# 4 Retrieve frozen sources and surface candidate passages
# Methodology: M2–M3. Research link: All RQs.
retrieval=[]; candidates=[]
if RUN_DOWNLOADS:
    from html import unescape
    for row in docs.to_dict('records'):
        try:
            if row['local_path']:
                payload=Path(row['local_path']).read_bytes()
            else:
                if not row['url'].startswith('https://'): raise ValueError('HTTPS URL required')
                req=urllib.request.Request(row['url'],headers={'User-Agent':'Academic document analysis; public sources'})
                with urllib.request.urlopen(req,timeout=60) as response:
                    payload=response.read(40_000_001)
                if len(payload)>40_000_000: raise ValueError('Document exceeds 40 MB limit')
            digest=hashlib.sha256(payload).hexdigest()
            if row['sha256'] and row['sha256']!=digest: raise ValueError('Registered SHA256 mismatch')
            # Hash-derived filenames avoid path traversal from document IDs.
            dest=OUT/'sources'/digest
            dest.write_bytes(payload)
            if payload.startswith(b'%PDF'):
                from pypdf import PdfReader
                chunks=[(f'page {i+1}',p.extract_text() or '') for i,p in enumerate(PdfReader(io.BytesIO(payload)).pages)]
            else:
                raw=payload.decode('utf-8',errors='replace')
                raw=re.sub(r'<(script|style)\b[^>]*>.*?</\1>',' ',raw,flags=re.S|re.I)
                raw=unescape(re.sub('<[^>]+>',' ',raw))
                chunks=[('text line '+str(i+1),t) for i,t in enumerate(raw.splitlines())]
            for locator,txt in chunks:
                for m in re.finditer(r'.{0,180}\b(?:appeal|deactivat\w*|human review|reinstat\w*|representativ\w*|discretion|complaint|suspend\w*)\b.{0,180}',txt,re.I):
                    candidates.append({'doc_id':row['doc_id'],'locator':locator,'candidate':m.group(0),'coding_status':'UNREVIEWED'})
            retrieval.append({'doc_id':row['doc_id'],'sha256':digest,'status':'saved','error':''})
        except Exception as e:
            retrieval.append({'doc_id':row['doc_id'],'sha256':'','status':'failed','error':str(e)})
pd.DataFrame(retrieval,columns=['doc_id','sha256','status','error']).to_csv(OUT/'audit'/'retrieval.csv',index=False)
candidate_df=pd.DataFrame(candidates,columns=['doc_id','locator','candidate','coding_status'])
candidate_df.to_csv(OUT/'audit'/'candidate_passages_UNREVIEWED.csv',index=False)
display(candidate_df.head(20))
print('Retrieval disabled; use registered local corpus.' if not RUN_DOWNLOADS else 'Review retrieval.csv for failures; do not code inaccessible documents as absent.')
```

```python
import pymupdf
import pandas as pd

pdf_path = "/kaggle/working/project12_sources/Fairwork_Cloudwork_Ratings_2023.pdf"

keywords = [
    "appeal",
    "dispute",
    "explanation",
    "human review",
    "representation",
    "deactivation",
    "complaint",
    "grievance"
]

matches = []

with pymupdf.open(pdf_path) as pdf:
    for page_num, page in enumerate(pdf, start=1):
        page_text = page.get_text()
        lower_text = page_text.lower()

        for keyword in keywords:
            start = 0
            while True:
                pos = lower_text.find(keyword, start)
                if pos == -1:
                    break

                excerpt = page_text[
                    max(0, pos - 150):min(len(page_text), pos + 250)
                ].replace("\n", " ")

                matches.append({
                    "page": page_num,
                    "keyword": keyword,
                    "excerpt": excerpt
                })

                start = pos + len(keyword)

matches_df = pd.DataFrame(
    matches,
    columns=["page", "keyword", "excerpt"]
)

output_path = (
    "/kaggle/working/project12_results/"
    "candidate_passages.csv"
)

matches_df.to_csv(output_path, index=False)

print("Candidate matches found:", len(matches_df))
print("Saved to:", output_path)

display(matches_df.head(15))
```

### 5 Validate the evidence and analysis grid

**Methodology:** M5. **Research link:** All RQs.

Validation blocks analysis when coded data are incomplete or evidence is inconsistent. `not_stated` requires a documented search scope. `not_applicable` requires a rationale. Review is required for every included case × domain × variable, including explicit inapplicability. Missing data never become zeros.


```python
# 5 Validate the evidence and analysis grid
# Methodology: M5. Research link: All RQs.
errors=[]
def require(test,message):
    if not test: errors.append(message)
def dates_ok(frame,cols,label):
    for col in cols:
        require(pd.to_datetime(frame[col],errors='coerce').notna().all(),f'{label}.{col}: valid ISO dates required')
    if len(frame) and not errors:
        require((pd.to_datetime(frame[cols[0]])<=pd.to_datetime(frame[cols[1]])).all(),f'{label}: reversed interval')
if len(codes):
    for name,frame,key in [('cases',cases,['platform_id']),('documents',docs,['doc_id']),('evidence',ev,['evidence_id']),
                           ('codes',codes,['platform_id','domain','variable']),('fairwork_matches',matches,['platform_id','domain','safeguard'])]:
        require(not frame.duplicated(key).any(),f'{name}: duplicate key {key}')
        require(not frame[key].eq('').any().any(),f'{name}: empty key')
    require(cases.included.isin(['0','1']).all(),'cases.included must be 0 or 1')
    require(cases.loc[cases.included.eq('0'),'exclusion_reason'].ne('').all(),'Excluded cases need reasons')
    included=cases[cases.included.eq('1')].copy()
    ids=set(included.platform_id)
    require(bool(ids),'No included cases')
    require(included[['platform','jurisdiction','work_category']].ne('').all().all(),'Case metadata is incomplete')
    require(docs.platform_id.isin(set(cases.platform_id)|{'GLOBAL'}).all(),'Unknown document case')
    require(docs.source_type.isin(['policy','fairwork','benchmark']).all(),'Unknown source_type')
    require(docs[['title','accessed_date','sha256']].ne('').all().all(),'Documents need title, retrieval date and frozen SHA256')
    require((docs.url.ne('')|docs.local_path.ne('')).all(),'Documents require a source URL or local path')
    require(docs.sha256.str.fullmatch('[a-fA-F0-9]{64}').all(),'Invalid SHA256')
    require(ev.doc_id.isin(docs.doc_id).all(),'Unknown evidence document')
    require(ev[['locator','excerpt_or_scope']].ne('').all().all(),'Evidence needs locators and excerpts/search scope')
    require(codes.platform_id.isin(ids).all(),'Final codes must reference included cases only')
    require(codes.value.isin(VALUES[:-1]).all(),'Unresolved or unknown codes must be adjudicated')
    require(codes[['coder_id','evidence_id']].ne('').all().all(),'Final codes need coder and evidence IDs')
    require(codes.evidence_id.isin(ev.evidence_id).all(),'Unknown code evidence')
    require(codes.loc[codes.value.eq('not_applicable'),'rationale'].ne('').all(),'Not-applicable codes require rationale')
    expected={(p,d,v) for p in ids for d in DOMAINS for v in DOMAIN_VARS}|{(p,'platform',v) for p in ids for v in GOV}
    observed=set(codes[['platform_id','domain','variable']].itertuples(index=False,name=None))
    require(expected==observed,f'Code grid mismatch: {len(expected-observed)} missing; {len(observed-expected)} unexpected')
    dates_ok(cases,['period_start','period_end'],'cases')
    dates_ok(docs,['valid_from','valid_to'],'documents')
    require(pd.to_datetime(docs.accessed_date,errors='coerce').notna().all(),'Invalid accessed_date')
    if not errors:
        freeze=pd.Timestamp(FREEZE_DATE)
        require((pd.to_datetime(docs.valid_to)<=freeze).all(),'Source validity extends beyond freeze; truncate at verified freeze')
        require((pd.to_datetime(cases.period_end)<=freeze).all(),'Case period extends beyond freeze')
        dx=docs.set_index('doc_id'); ex=ev.set_index('evidence_id')
        for p in ids:
            types=set(docs.loc[docs.platform_id.eq(p),'source_type'])
            require({'policy','fairwork'}<=types,f'{p}: both policy and Fairwork sources needed')
        for r in codes.itertuples():
            source=dx.loc[ex.loc[r.evidence_id,'doc_id']]
            require(source.platform_id==r.platform_id and source.source_type=='policy',f'{r.platform_id}/{r.variable}: code must cite own policy')
            case=included.set_index('platform_id').loc[r.platform_id]
            require(max(pd.Timestamp(source.valid_from),pd.Timestamp(case.period_start))<=min(pd.Timestamp(source.valid_to),pd.Timestamp(case.period_end)),f'{r.platform_id}: policy outside case period')
        claimkeys=set(codes.loc[codes.variable.isin(SAFE)&codes.value.eq('present'),['platform_id','domain','variable']].itertuples(index=False,name=None))
        matchkeys=set(matches[['platform_id','domain','safeguard']].itertuples(index=False,name=None))
        require(claimkeys==matchkeys,'Fairwork matches must cover exactly all positive domain safeguard claims')
        require(matches.status.isin(STATUS).all(),'Invalid Fairwork comparison status')
        require(matches.rationale.ne('').all(),'Fairwork match needs interpretive rationale, including scoped silence')
        require(matches.evidence_id.isin(ev.evidence_id).all(),'Unknown Fairwork evidence')
        require(matches.policy_doc_id.isin(docs.doc_id).all() and matches.fairwork_doc_id.isin(docs.doc_id).all(),'Unknown match documents')
        if not errors:
            keyed=codes.set_index(['platform_id','domain','variable'])
            for r in matches.itertuples():
                a,b=dx.loc[r.policy_doc_id],dx.loc[r.fairwork_doc_id]
                claim=keyed.loc[(r.platform_id,r.domain,r.safeguard)]
                claimdoc=ex.loc[claim.evidence_id,'doc_id']
                require(a.source_type=='policy' and b.source_type=='fairwork','Match source types incorrect')
                require(a.platform_id==b.platform_id==r.platform_id,'Match case/jurisdiction mismatch')
                require(claimdoc==r.policy_doc_id,'Match policy is not the actual claim source')
                require(ex.loc[r.evidence_id,'doc_id']==r.fairwork_doc_id,'Independent evidence not from matched report')
    pd.DataFrame({'error':errors}).to_csv(OUT/'audit'/'validation_errors.csv',index=False)
    if errors: raise ValueError('Validation failed:\n'+'\n'.join(errors[:40]))
    READY=True
    codes=codes.merge(ev[['evidence_id','doc_id']],on='evidence_id',validate='many_to_one')
    wide=codes.pivot(index=['platform_id','domain'],columns='variable',values='value')
    domain=wide.loc[wide.index.get_level_values('domain').isin(DOMAINS)].copy()
    gov=wide.xs('platform',level='domain')[GOV].copy()
    print(f'VALIDATED: {len(ids)} cases, {len(codes)} codes, {len(matches)} positive claims.')
else:
    print('AWAITING_CODED_DATA: templates available; all 20 empirical outputs are intentionally withheld.')
```

### 7 Define publication exports and denominators

**Methodology:** M6 and M9. **Research link:** All RQs.

Every requested figure is displayed and saved as vector PDF plus PNG; its data are saved as CSV. Every table is displayed in full and saved as CSV. Captions define denominators and interpretation. Display size adapts to the corpus; do not infer real-world absence from `not_stated`.


```python
# 7 Define publication exports and denominators
# Methodology: M6 and M9. Research link: All RQs.
CAPTIONS={}
def table(name,df,caption):
    df=df.reset_index(drop=True)
    df.to_csv(OUT/'tables'/f'{name}.csv',index=False)
    CAPTIONS[name]=caption
    ARTIFACTS.append({'id':name,'type':'table','rows':len(df),'caption':caption})
    display(Markdown(f'**{name}.** {caption}'))
    with pd.option_context('display.max_rows',None,'display.max_columns',None,'display.max_colwidth',100): display(df)
def figure(name,fig,df,caption):
    CAPTIONS[name]=caption
    fig.savefig(OUT/'figures'/f'{name}.pdf',bbox_inches='tight')
    fig.savefig(OUT/'figures'/f'{name}.png',dpi=300,bbox_inches='tight')
    df.to_csv(OUT/'tables'/f'{name}_plot_data.csv',index=False)
    ARTIFACTS.append({'id':name,'type':'figure','rows':len(df),'caption':caption})
    display(Markdown(f'**{name}.** {caption}'))
    plt.show(); plt.close(fig)
def presence(frame,columns):
    applicable=frame[columns].ne('not_applicable').sum(axis=1)
    count=frame[columns].eq('present').sum(axis=1)
    return count.div(applicable.replace(0,np.nan)), count, applicable
LABELS={'present':'P','not_stated':'NS','not_applicable':'NA'}
def signature(frame,columns):
    return frame[columns].apply(lambda r:' / '.join(LABELS[v] for v in r),axis=1)
def heatmap(name,frame,caption):
    data=frame.apply(lambda col: col.map({'present':2,'not_stated':1,'not_applicable':0})).astype(float)
    fig,ax=plt.subplots(figsize=(max(6,len(data.columns)*1.1),max(3,len(data)*.21+1.6)))
    cmap=ListedColormap(['#cccccc','#f3ead2','#246878'])
    im=ax.imshow(data.values,aspect='auto',cmap=cmap,norm=BoundaryNorm([-.5,.5,1.5,2.5],3))
    ax.set_xticks(range(len(data.columns)),[c.replace('_',' ') for c in data.columns],rotation=35,ha='right')
    ax.set_yticks(range(len(data)),[' | '.join(map(str,i)) if isinstance(i,tuple) else str(i) for i in data.index],fontsize=7)
    cb=fig.colorbar(im,ax=ax,ticks=[0,1,2],shrink=.4)
    cb.ax.set_yticklabels(['Not applicable','Not stated','Present'])
    figure(name,fig,frame.reset_index(),caption)
def stacked(name,df,title,caption):
    fig,ax=plt.subplots(figsize=(8,max(3,len(df)*.4)))
    df.plot.barh(stacked=True,ax=ax,color=['#246878','#e6ad50','#a74648','#b5bbc0'][:len(df.columns)])
    ax.set_xlabel('Number of cases or claims'); ax.set_title(title)
    ax.legend(bbox_to_anchor=(1.02,1),loc='upper left',frameon=False)
    figure(name,fig,df.reset_index(),caption)
def annotate_codes(frame):
    keys=frame[['platform_id','domain']].drop_duplicates()
    selected=codes.merge(keys,on=['platform_id','domain'])
    return selected[['platform_id','domain','variable','value','evidence_id','doc_id','rationale']]
if READY:
    domain['protection'],domain['protection_n'],domain['protection_denom']=presence(domain,CORE)
    domain['discretion'],domain['discretion_n'],domain['discretion_denom']=presence(domain,POWERS)
    domain['gap']=domain.discretion-domain.protection
    domain['signature']=signature(domain,CORE)
    domain.to_csv(OUT/'audit'/'domain_analysis.csv')
```

### 10 RQ3 Triangulate policy commitments with Fairwork

**Methodology:** M5 and M7. **Research link:** RQ3.

Only explicit positive safeguard commitments enter the comparison. Interval overlap is the primary contemporaneity criterion. Sensitivity windows admit progressively more distant policy/report periods. Contradiction share excludes report silence and must always display its discussed-claim denominator.


```python
# 10 RQ3 Triangulate policy commitments with Fairwork
# Methodology: M5 and M7. Research link: RQ3.
if READY:
    comp=matches.copy()
    dx=docs.set_index('doc_id')
    def gap_days(r):
        a,b=dx.loc[r.policy_doc_id],dx.loc[r.fairwork_doc_id]
        start=max(pd.Timestamp(a.valid_from),pd.Timestamp(b.valid_from))
        end=min(pd.Timestamp(a.valid_to),pd.Timestamp(b.valid_to))
        return max(0,(start-end).days)
    comp['temporal_gap_days']=comp.apply(gap_days,axis=1) if len(comp) else pd.Series(dtype=float)
    comp['primary_eligible']=comp.temporal_gap_days.le(PRIMARY_WINDOW_DAYS)
    comp=comp.merge(ev.rename(columns={'doc_id':'independent_evidence_doc_id'}),on='evidence_id',validate='many_to_one')
    eligible=comp[comp.primary_eligible]
    counts=pd.crosstab(eligible.safeguard,eligible.status).reindex(index=SAFE,columns=STATUS,fill_value=0)
    stacked('RQ3_F1',counts,'Independent evidence for positive policy claims','Counts for temporally eligible positive claims only. Not discussed is report silence; it is neither corroboration nor contradiction.')
    sensitivity=[]
    for window in [0,90,365]:
        sub=comp[comp.temporal_gap_days.le(window)]
        discussed=sub.status.ne('not_discussed').sum()
        sensitivity.append({'window_days':window,'all_positive_claims':len(comp),'eligible_claims':len(sub),'discussed_claims':discussed,'contradicted_claims':sub.status.eq('contradicted').sum(),'matching_coverage':len(sub)/len(comp) if len(comp) else np.nan,'contradiction_share':sub.status.eq('contradicted').sum()/discussed if discussed else np.nan})
    sens=pd.DataFrame(sensitivity)
    fig,axes=plt.subplots(1,2,figsize=(9,3.8))
    for ax,col,label in zip(axes,['matching_coverage','contradiction_share'],['Temporally matched / all positive claims','Contradicted / discussed matched claims']):
        ax.plot(sens.window_days,sens[col],marker='o',color='#246878'); ax.set(xlabel='Maximum interval gap in days',ylabel=label,ylim=(-.05,1.05)); ax.set_xticks([0,90,365])
    fig.tight_layout()
    figure('RQ3_F2',fig,sens,'Temporal sensitivity. Zero days means overlap; 90/365 days are relaxed comparisons, not proof of contemporaneity. Undefined proportions remain missing.')
    table('RQ3_T1',comp,'Full positive-claim evidence crosswalk, including unmatched pairs and temporal eligibility. Source dates and URLs are in the document register.')
    total=codes[codes.variable.isin(SAFE)&codes.value.eq('present')].groupby('variable').size().reindex(SAFE,fill_value=0)
    tab=counts.copy(); tab['all_positive_claims']=total; tab['eligible_claims']=counts.sum(axis=1); tab['discussed_claims']=counts[['corroborated','qualified','contradicted']].sum(axis=1)
    table('RQ3_T2',tab.rename_axis('safeguard').reset_index(),'Safeguard-level triangulation with all-claim, matched-claim and discussed-claim denominators. A Fairwork score alone is never coded as a contradiction.')
else: print('RQ3 outputs await validated codes.')
```

### 12 RQ5 Distinguish individual and collective governance rights

**Methodology:** M7. **Research link:** RQ5.

Consultation and binding influence have distinct codes. The configuration crosswalk is descriptive and does not imply that representative rights cause individual protection. Platform-level entitlement codes must be read alongside domain coverage.


```python
# 12 RQ5 Distinguish individual and collective governance rights
# Methodology: M7. Research link: RQ5.
if READY:
    heatmap('RQ5_F1',gov,'Individual appeal and collective governance rights by case. Representative assistance in an individual dispute is distinct from participation in revising rules.')
    ct=pd.crosstab(gov.individual_challenge,gov.rule_influence).reindex(index=VALUES[:-1],columns=VALUES[:-1],fill_value=0)
    stacked('RQ5_F2',ct,'Individual challenge and representative rule influence','Case counts by individual challenge status (rows) and binding representative rule influence (legend). The plot does not presume a universal ordinal ladder.')
    linked=codes[codes.domain.eq('platform')&codes.variable.isin(GOV)][['platform_id','variable','value','evidence_id','doc_id','rationale']]
    table('RQ5_T1',linked,'Governance rights evidence crosswalk. Rule consultation does not satisfy the stricter rule-influence definition.')
    complete=domain[CORE].eq('present').all(axis=1).groupby(level='platform_id').any().rename('complete_bundle_any_domain')
    cross=gov[['rule_influence']].join(complete)
    counts=cross.groupby(['rule_influence','complete_bundle_any_domain']).size().rename('cases').reset_index()
    table('RQ5_T2',counts,'Representative rule influence versus a complete four-safeguard bundle in at least one decision domain. Sparse cells are descriptive; no inference or causal interpretation.')
else: print('RQ5 outputs await validated codes.')
```
